Q11. Print all prime numbers in a range entered by the user

````python
lb = int(input("Enter lower bound: "))
hb = int(input("Enter higher bound: "))

def check_prime(n):
  if n < 2: return False
  for i in range(2, int(math.sqrt(n)) + 1): # Check factors up to square root
    if n % i == 0: return False
  return True

# Filter numbers in range and print as a list
print([x for x in range(lb, hb + 1) if check_prime(x)])
````

Q14. Display whether a year is a leap year or not

````python
def check_leap_year(x):
  # Leap year if divisible by 4 AND (not by 100 OR by 400)
  return (x % 4 == 0 and x % 100 != 0) or (x % 400 == 0)

x = int(input("Enter the year to be checked: "))
print(f"{x} is a leap year" if check_leap_year(x) else f"{x} is not a leap year")
````

Q21. Write a program to reverse a number using loops

````python
def reverse_number(x):
  ans = 0
  while(x > 0):
    ans = (ans * 10) + (x % 10) # Append last digit of x to ans
    x //= 10 # Remove last digit
  return ans

print(reverse_number(int(input("Enter number to be reversed: "))))
````
