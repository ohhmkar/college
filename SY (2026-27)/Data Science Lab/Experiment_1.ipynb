Q11. Print all prime numbers in a range entered by the user

````python
lb = int(input("Enter lower bound of the program: "))
hb = int(input("Enter higher bound of the program: "))

def check_prime(x):
  for i in range(2, int(x**0.5) + 1):
    if x % i == 0:
      return False
  return x > 1

print(*[x for x in range(lb, hb) if check_prime(x)])
````

Q14. Display whether a year is a leap year or not

````python
def check_leap_year(x):
  return True if (x>0 and x%4 == 0) else False

x = int(input("Enter the year to be checked: "))
print(f"{x} is a leap year" if check_leap_year(x) else f"{x} is not a leap year")
````

Q19. Write a simple calculator program using if-elif-else

````python
x, y = map(int, input("Enter 2 numbers: ").split())
choice = input(f"What operation would you like to perform with {x} and {y}: ")

if choice == "Add":
  print(f"Sum of {x} and {y} is: {x + y}")
elif choice == "Divide":
  print(f"Quotient of {x} and {y} is: {x / y}")
elif choice == "Subtract":
  print(f"Difference of {x} and {y} is: {x - y}")
elif choice == "Multiply":
  print(f"Product of {x} and {y} is: {x * y}")
else:
  print("Invalid choice")
````

Q21. Write a program to reverse a number using loops

````python
def reverse_number(x):
  ans = 0
  while(x>0):
    ans*=10
    ans += x%10
    x//=10
  return ans

print(reverse_number(int(input("Enter number to be reversed: "))))

````
